```python
import os
import json
import pandas as pd
import faiss
import numpy as np
from sentence_transformers import SentenceTransformer, util

MODEL_NAME = 'sentence-transformers/paraphrase-multilingual-mpnet-base-v2'
model = SentenceTransformer(MODEL_NAME)
```

```python
FOLDER = r"D:\finalproject\KLTN\Backend\data\vector_store"

K_LIST = [5, 10, 20]

SEMANTIC_THRESHOLD = 0.72

store_pairs = []

for filename in os.listdir(FOLDER):
    if filename.startswith("documents_") and filename.endswith(".json"):
        store_name = filename[len("documents_"):-len(".json")]

        json_file = os.path.join(FOLDER, filename)
        faiss_file = os.path.join(FOLDER, f"faiss_index_{store_name}.bin")

        if os.path.exists(faiss_file):
            store_pairs.append((store_name, json_file, faiss_file))
```

```python
all_store_data = {}

for store_name, json_file, faiss_file in store_pairs:
    with open(json_file, 'r', encoding='utf-8') as f:
        data = json.load(f)

    chunks = [doc.get('content', '') for doc in data]

    index = faiss.read_index(faiss_file)

    all_store_data[store_name] = {
        "chunks": chunks,
        "index": index
    }

precomputed_embeddings = {}

for name, store in all_store_data.items():
    chunks = store["chunks"]

    if len(chunks) == 0:
        dim = model.get_sentence_embedding_dimension()
        precomputed_embeddings[name] = np.zeros((0, dim), dtype=np.float32)
    else:
        embs = model.encode(
            chunks,
            normalize_embeddings=True,
            convert_to_numpy=True,
            show_progress_bar=False
        )
        precomputed_embeddings[name] = embs

def compute_ndcg(retrieved_sims, threshold):
    relevance = [1 if sim >= threshold else 0 for sim in retrieved_sims]

    def dcg(scores):
        return sum(score / np.log2(i + 2) for i, score in enumerate(scores))

    ideal = sorted(relevance, reverse=True)
    denom = dcg(ideal) if sum(ideal) > 0 else 1.0

    return dcg(relevance) / (denom + 1e-10)
```

```python
import numpy as np
from sentence_transformers import util

# Tổng số câu hỏi
total_queries = len(GROUND_TRUTH_QA)
results = {}

print(f"{'K':>4} | {'Queries':>7} | {'Recall@K':>8} | {'MRR':>6} | "
      f"{'Precision@K':>12} | {'NDCG@K':>8} | {'AvgRank1':>9} | {'Coverage':>8}")
print("-" * 90)

for K in K_LIST:

    # Các metric cần tính
    recall_scores = []
    hit_at_k_scores = []
    mrr_scores = []
    precision_scores = []
    ndcg_scores = []
    avg_rank_scores = []
    coverage_count = 0

    k_per_store = K   # số kết quả lấy từ mỗi store

    # ----------------------------------------------------------
    # Duyệt từng câu hỏi trong Ground Truth
    # ----------------------------------------------------------
    for qa in GROUND_TRUTH_QA:
        query = qa["question"]
        gt_chunks = qa["ground_truth_chunks"]

        # Encode query và ground truth
        query_emb = model.encode([query], normalize_embeddings=True, convert_to_numpy=True)
        gt_embs = model.encode(gt_chunks, normalize_embeddings=True, convert_to_numpy=True)

        # -----------------------------
        # Lấy candidates từ tất cả store
        # -----------------------------
        candidates = []

        for store_name, store in all_store_data.items():
            index = store["index"]

            # Lấy số vector trong FAISS index
            ntotal = getattr(index, "ntotal", None)
            if ntotal is None:
                try:
                    ntotal = index.ntotal
                except Exception:
                    ntotal = 0

            # Nếu store không có dữ liệu thì bỏ qua
            k_search = min(k_per_store, ntotal) if ntotal > 0 else 0
            if k_search == 0:
                continue

            # Tìm kiếm trên FAISS
            D, I = index.search(np.array(query_emb, dtype=np.float32), k_search)

            # Gộp candidates
            for dist, idx in zip(D[0], I[0]):
                if idx >= 0:
                    emb = precomputed_embeddings[store_name][idx]
                    text = store["chunks"][idx]
                    candidates.append((emb, text))

        # Trường hợp không có candidates
        if len(candidates) == 0:
            recall_scores.append(0.0)
            hit_at_k_scores.append(0.0)
            mrr_scores.append(0.0)
            precision_scores.append(0.0)
            ndcg_scores.append(0.0)
            avg_rank_scores.append(K)
            continue

        # ----------------------------------------------------------
        # Tính cosine similarity và sắp xếp theo độ liên quan
        # ----------------------------------------------------------
        sims_with_cands = []
        for emb, text in candidates:
            sim = util.cos_sim(query_emb, emb).item()
            sims_with_cands.append((sim, emb, text))

        sims_with_cands.sort(key=lambda x: -x[0])
        ranked = sims_with_cands[:K]

        ranked_embs = [r[1] for r in ranked]

        # ----------------------------------------------------------
        # Tính similarity với ground-truth chunks
        # ----------------------------------------------------------
        semantic_sims = []
        for emb in ranked_embs:
            sims = util.cos_sim(np.array([emb]), gt_embs)
            max_sim = float(sims.max().item())
            semantic_sims.append(max_sim)

        # ----------------------------------------------------------
        # Tính các metrics
        # ----------------------------------------------------------

        # Hit/Recall@K
        hit = any(sim >= SEMANTIC_THRESHOLD for sim in semantic_sims)
        recall_scores.append(1.0 if hit else 0.0)
        hit_at_k_scores.append(1.0 if hit else 0.0)

        # MRR
        rr = 0.0
        for idx, sim in enumerate(semantic_sims, start=1):
            if sim >= SEMANTIC_THRESHOLD:
                rr = 1.0 / idx
                break
        mrr_scores.append(rr)

        # Precision@K
        prec = sum(1 for sim in semantic_sims if sim >= SEMANTIC_THRESHOLD) / len(semantic_sims)
        precision_scores.append(prec)

        # NDCG@K
        ndcg_scores.append(compute_ndcg(semantic_sims, SEMANTIC_THRESHOLD))

        # Avg rank của hit đầu tiên
        if rr > 0:
            avg_rank_scores.append(1.0 / rr)
        else:
            avg_rank_scores.append(K)

        coverage_count += 1

    # ----------------------------------------------------------
    # Tổng hợp kết quả theo K
    # ----------------------------------------------------------
    results[K] = {
        "TotalQueries": coverage_count,
        "Recall": float(np.mean(recall_scores)) if recall_scores else 0.0,
        "Hit": float(np.mean(hit_at_k_scores)) if hit_at_k_scores else 0.0,
        "MRR": float(np.mean(mrr_scores)) if mrr_scores else 0.0,
        "Precision": float(np.mean(precision_scores)) if precision_scores else 0.0,
        "NDCG": float(np.mean(ndcg_scores)) if ndcg_scores else 0.0,
        "AvgRankFirstHit": float(np.mean(avg_rank_scores)) if avg_rank_scores else float(K),
        "Coverage": float(coverage_count) / max(1, total_queries)
    }

    # Print kết quả từng K
    r = results[K]
    print(f"{K:>4} | {r['TotalQueries']:7d} | {r['Recall']:.4f} | {r['MRR']:.4f} | "
          f"{r['Precision']:.4f} | {r['NDCG']:.4f} | {r['AvgRankFirstHit']:.2f} | {r['Coverage']:.3f}")
```

```python
def evaluate_single_case(query, gt_chunks, K=10):

    query_emb = model.encode([query], normalize_embeddings=True, convert_to_numpy=True)
    gt_embs = model.encode(gt_chunks, normalize_embeddings=True, convert_to_numpy=True)

    candidates = []

    # Lấy candidates từ tất cả store (giống code bạn đang dùng)
    for store_name, store in all_store_data.items():
        index = store["index"]

        ntotal = getattr(index, "ntotal", 0)
        if ntotal <= 0:
            continue

        k_search = min(K, ntotal)
        D, I = index.search(np.array(query_emb, dtype=np.float32), k_search)

        for dist, idx in zip(D[0], I[0]):
            if idx >= 0:
                candidates.append((
                    precomputed_embeddings[store_name][idx],
                    store["chunks"][idx],
                ))

    if len(candidates) == 0:
        return {
            "hit": 0,
            "recall": 0,
            "mrr": 0,
            "precision": 0,
            "ndcg": 0,
            "rank1": K
        }

    sims_with_cands = []
    for emb, text in candidates:
        sim = util.cos_sim(query_emb, emb).item()
        sims_with_cands.append((sim, emb, text))

    sims_with_cands.sort(key=lambda x: -x[0])
    ranked = sims_with_cands[:K]
    ranked_embs = [r[1] for r in ranked]

    semantic_sims = []
    for emb in ranked_embs:
        sims = util.cos_sim(np.array([emb]), gt_embs)
        max_sim = float(sims.max().item())
        semantic_sims.append(max_sim)

    # Hit
    hit = any(sim >= SEMANTIC_THRESHOLD for sim in semantic_sims)

    # MRR
    rr = 0
    for i, sim in enumerate(semantic_sims, start=1):
        if sim >= SEMANTIC_THRESHOLD:
            rr = 1 / i
            break

    # Precision
    precision = sum(1 for s in semantic_sims if s >= SEMANTIC_THRESHOLD) / len(semantic_sims)

    # NDCG
    ndcg = compute_ndcg(semantic_sims, SEMANTIC_THRESHOLD)

    # Rank1
    rank1 = (1 / rr) if rr > 0 else K

    return {
        "hit": int(hit),
        "recall": int(hit),
        "mrr": rr,
        "precision": precision,
        "ndcg": ndcg,
        "rank1": rank1
    }
```

```python
all_results = []

for case in EDGE_CASE_QA:

    result = evaluate_single_case(
        query=case["question"],
        gt_chunks=case["ground_truth_chunks"],
        K=10
    )

    # Thêm metadata để lát in bảng
    result["category"] = case["category"]
    result["question"] = case["question"]

    all_results.append(result)

# ============================
# IN BẢNG KẾT QUẢ ĐẦY ĐỦ
# ============================

df = pd.DataFrame(all_results)
```
